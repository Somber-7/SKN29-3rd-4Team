# Ollama + LangChain 연동 테스트

로컬 Ollama 서버(qwen3.5:4b)와 LangChain ChatOllama 연결을 확인합니다.

```python
import ollama
from langchain_ollama import ChatOllama

client = ollama.Client(
    host="http://localhost:11434",
    timeout=300,
)

llm = ChatOllama(
    model="qwen3.5:4b",
    client=client,
    temperature=0.7,
    num_ctx=32768,
)
```

```python
# 기본 연결 확인
response = llm.invoke("안녕하세요. 한 문장으로 자기소개 해주세요. /no_think")
print(response.content)
```

```python
# 작명 도메인 기본 응답 확인
response = llm.invoke(
    "'김'씨 성을 가진 아이의 이름을 지을 때 木 오행에 맞는 한자 이름 2개만 추천해줘. /no_think"
)
print("content:", repr(response.content))
print("thinking:", repr(response.additional_kwargs.get("thinking", ""))[:100])
print("---")
print(response.content)
```

```python
# 스트리밍 확인
for chunk in llm.stream("'이'씨 성에 어울리는 순우리말 이름 2개 추천해줘. /no_think"):
    print(chunk.content, end="", flush=True)
```
